# Market data exploratory analysis
**Simulation only. NOT investment advice. Virtual money, no real brokerage connection.**
The supplied source is synthetic and includes weekends. The figures use cleaned supplied prices. Market history, when available, is stored separately in `data/market_cache.csv`.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
prices = pd.read_csv(root / 'data' / 'cleaned_dataset.csv', parse_dates=['date'])
prices.head()
```

## 1. Closing-price trends

```python
px.line(prices, x='date', y='close', color='ticker', title='Closing prices by ticker').show()
```

## 2. Daily-return distribution

```python
px.histogram(prices.dropna(subset=['daily_return_pct']), x='daily_return_pct', color='ticker', nbins=45, title='Close-to-close daily return (%)').show()
```

## 3. Return correlation

```python
returns = prices.pivot(index='date', columns='ticker', values='daily_return_pct')
px.imshow(returns.corr(), zmin=-1, zmax=1, color_continuous_scale='RdBu', title='Return correlation').show()
```

## 4. Trading volume

```python
px.box(prices, x='ticker', y='volume', title='Volume distribution by ticker').show()
```
